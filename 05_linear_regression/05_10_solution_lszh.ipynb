# TSO Semester Week 5: Linear Regression
In this exercise, we will focus on linear regression models using Python's `statsmodels` library. We will go through the steps of loading data, preprocessing, training a linear regression model, and evaluating its performance. The first part of the exercise is to build a simple linear regression model, while the second part will extend this to a multiple linear regression model.


## Part 1: Simple Linear Regression Model
In the following, we will build a simple linear regression model to predict a target variable $y$ based on a single feature $X$. In order to do so, we will conduct the following steps:
1. Load the data into Python
2. Visualize the data to get a better understanding for the relationship between $X$ and $y$
3. Define the input and output variables of the linear regression model
4. Train the linear regression model
5. Inspect the model
6. Make predictions using the trained model
7. Visualize observations and predictions


### Part 1, Step 1 — Load the Data into Python

**Tasks:**
1. Load the dataset from the provided CSV file into a Pandas DataFrame.
2. Filter the dataset to include only rows in which feature `Year` is less than 2020 (this way, we exclude the COVID-19 pandemic years).

```python
df_full = pd.read_csv(path_to_lszh, delimiter=';')

# Filter the DataFrame for years before 2020
df = df_full.query("Year < 2020")
df
```

### Part 1, Step 2 — Visualize the Relationship between X and y

**Tasks:**
1. Create three different scatter plots to visualize the relationship between the features $X$ (`Year`, `Population`, `GDP_per_capita_USD`) and the target variable $y$ (`PAX_per_Year`).

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Year"],
        y=df["PAX_per_Year"],
        mode="markers",
        marker=dict(opacity=0.7),
        name="PAX"
    )
)

fig.update_layout(
    title="PAX per Year vs. Year",
    xaxis_title="Year",
    yaxis_title="PAX per Year",
    template="plotly_white"
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["GDP_per_capita_USD"],
        y=df["PAX_per_Year"],
        mode="markers",
        marker=dict(opacity=0.7),
        name="PAX"
    )
)

fig.update_layout(
    title="PAX per Year vs. GDP per capita USD",
    xaxis_title="GDP per capita USD",
    yaxis_title="PAX per Year",
    template="plotly_white"
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Population"],
        y=df["PAX_per_Year"],
        mode="markers",
        marker=dict(opacity=0.7),
        name="PAX"
    )
)

fig.update_layout(
    title="PAX per Year vs. Population",
    xaxis_title="Population",
    yaxis_title="PAX per Year",
    template="plotly_white"
)

fig.show()
```

### Part 1, Step 3 — Define Input (X) and Output (y)

**Tasks:**
1. Define the input variable `X` and the output variable `y` for the linear regression model. Only select one variable / feature for `X` (e.g., `Year`)


```python
feature = "Year"

X = df[feature]
y = df["PAX_per_Year"]
```

### Part 1, Step 4 — Train a Linear Regression Model

**Tasks:**
1. Use the `statsmodels` library to train a simple linear regression model using the defined input and output variables. Make sure to add the intercept term to $X$ before training the model.

```python
# statsmodels does NOT add an intercept automatically
X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()
```

### Part 1, Step 5 — Inspect the Model with `results.summary()` (statsmodels)

**Tasks:**
1. Use the `summary()` method of the fitted model to inspect the model's.
2. Evaluate the model-level performance metrics (i.e., R-squared, Adjusted R-squared, and F-statistic). What do these metrics indicate about the model's performance?
3. Evaluate the coefficient-level statistics (i.e., coefficients, standard errors, t-values, and p-values). What do these statistics indicate about the significance of the predictor variable?

```python
print(results.summary())
```

> ### 📊 Regression Model Summary
>
> **Model-level performance metrics**
>
> - **R-squared (0.934):**  
>   93.4 % of the variance in annual passenger numbers (`PAX_per_Year`) is explained by `Year`, indicating a **very strong linear relationship** between time and passenger demand.
>
> - **Adjusted R-squared (0.932):**  
>   Only marginally lower than the R-squared, as expected for a model with **one explanatory variable**, suggesting no overfitting.
>
> - **F-statistic (507.9):**  
>   The extremely small p-value allows us to **strongly reject the null hypothesis** that the slope coefficient equals zero, indicating the model is **highly statistically significant overall**.
>
> **Coefficient-level statistics**
>
> - **Intercept:**  
>   Represents the predicted passenger volume at `Year = 0`, which has **no meaningful real-world interpretation** in this context and is therefore not substantively relevant.
>
> - **Year:**  
>   Annual passenger numbers increase by approximately **571,000 passengers per year** on average. The large t-value, extremely small p-value, and narrow confidence interval provide **strong evidence that `Year` is a highly statistically significant predictor** of `PAX_per_Year`.

### Part 1, Step 6 — Make Predictions

**Tasks:**
1. Use the trained linear regression model to make predictions on the input data `X`. 
2. Add the predicted values to the original DataFrame as a new column named `predicted_PAX_per_Year`.

```python
df["predicted_PAX_per_Year"] = results.predict(X)
df.head(5)
```

### Part 1, Step 7 — Visualize Model vs Data

**Tasks:**
1. Create a scatter plot of the original data points (actual values) and overlay the regression line (predicted values) to visualize the fit of the model. Plot `Year` on the x-axis and both `PAX_per_Year` and `predicted_PAX_per_Year` on the y-axis.

```python
fig = go.Figure()

# Observed data points
fig.add_trace(
    go.Scatter(
        x=df[feature],
        y=df["PAX_per_Year"],
        mode="markers",
        name="Observed",
        marker=dict(
            symbol="diamond",
            color="red",
            size=8,
            opacity=0.8
        )
    )
)

# Regression line
fig.add_trace(
    go.Scatter(
        x=df[feature],
        y=df["predicted_PAX_per_Year"],
        mode="lines",
        name="Linear Regression Model",
        line=dict(
            color="blue",
            width=2
        )
    )
)

fig.update_layout(
    title="",
    xaxis_title=feature,
    yaxis_title="Annual Passengers Zurich Airport",
    template="plotly_white",
    legend=dict(
        font=dict(size=12)
    )
)

fig.update_xaxes(title_font=dict(size=14))
fig.update_yaxes(title_font=dict(size=14))

fig.show()
```

# Multiple Linear Regression Model

```python
features = ["Year", "GDP_per_capita_USD", "Population"]

X = df[features]
y = df["PAX_per_Year"]
```

```python
# statsmodels does NOT add an intercept automatically
X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()
```

```python
print(results.summary())
```

```python
df["predicted_PAX_per_Year"] = results.predict(X)
```

```python
df_pred = df_full.query("Year >= 2020").copy() 
df_pred
```

```python
X_pred = df_pred[features]
X_pred = sm.add_constant(X_pred)
df_pred["predicted_PAX_per_Year"] = results.predict(X_pred)
df_pred
```

```python
fig = go.Figure()

# Observed data points (pre-2020)
fig.add_trace(
    go.Scatter(
        x=df["Year"],
        y=df["PAX_per_Year"],
        mode="markers",
        name="Observed (pre-2020)",
        marker=dict(
            symbol="diamond",
            color="red",
            size=8,
            opacity=0.8
        )
    )
)

# Observed data points (2020+)
fig.add_trace(
    go.Scatter(
        x=df_pred["Year"],
        y=df_pred["PAX_per_Year"],
        mode="markers",
        name="Observed (2020+)",
        marker=dict(
            symbol="diamond",
            color="red",
            size=8,
            opacity=0.8
        )
    )
)

# Observed data points (2020+)
fig.add_trace(
    go.Scatter(
        x=df_pred["Year"],
        y=df_pred["predicted_PAX_per_Year"],
        mode="lines",
        name="Predicted (2020+)",
        line=dict(
            color="green",
            width=2
        )
    )
)

# Regression line
fig.add_trace(
    go.Scatter(
        x=df["Year"],
        y=df["predicted_PAX_per_Year"],
        mode="lines",
        name="Linear Regression Model",
        line=dict(
            color="blue",
            width=2
        )
    )
)

fig.update_layout(
    title="",
    xaxis_title="Year",
    yaxis_title="Annual Passengers Zurich Airport",
    template="plotly_white",
    legend=dict(
        font=dict(size=12)
    )
)

fig.update_xaxes(title_font=dict(size=14))
fig.update_yaxes(title_font=dict(size=14))

fig.show()
```
